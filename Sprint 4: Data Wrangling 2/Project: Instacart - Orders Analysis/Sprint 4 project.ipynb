    - `'order_id'`: número de ID que identifica de manera única cada pedido.
    - `'product_id'`: número ID que identifica de manera única cada producto.
    - `'add_to_cart_order'`: el orden secuencial en el que se añadió cada artículo en el carrito.

- `aisles.csv`
    - `'aisle_id'`: número ID que identifica de manera única cada categoría de pasillo de víveres.
    - `'aisle'`: nombre del pasillo.
- `departments.csv`
    - `'department_id'`: número ID que identifica de manera única cada departamento de víveres.
    - `'department'`: nombre del departamento.

### [C3] Para cada producto, ¿cuál es la proporción de las veces que se pide y que se vuelve a pedir?

```python
# Contar cuántas veces se pidió cada producto
total_orders = order_products['product_id'].value_counts().reset_index()
total_orders.columns = ['product_id', 'total_orders']

# Contar cuántas veces se reordenó cada producto
reorder_counts = (
    order_products[order_products['reordered'] >= 1]['product_id']
    .value_counts()
    .reset_index()
)
reorder_counts.columns = ['product_id', 'reorder_count']

# Combinar los datos
product_reorder_ratio = pd.merge(
    total_orders,
    reorder_counts,
    on='product_id',
    how='left'
)

```

```python
# Rellenar valores NaN (productos nunca reordenados)
product_reorder_ratio['reorder_count'] = product_reorder_ratio['reorder_count'].fillna(0)

# Calcular la proporción
product_reorder_ratio['reorder_ratio'] = (
    product_reorder_ratio['reorder_count'] / product_reorder_ratio['total_orders']
)

# Agregar los nombres de los productos
product_reorder_ratio = pd.merge(
    product_reorder_ratio,
    products[['product_id', 'product_name']],
    on='product_id',
    how='left'
)

# Ordenar por proporción descendente
product_reorder_ratio = product_reorder_ratio.sort_values(by='reorder_ratio', ascending=False)
#product_reorder_ratio_1 = product_reorder_ratio[product_reorder_ratio['reorder_ratio']==0]
#print(product_reorder_ratio)
product_reorder_ratio.describe()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(
    product_reorder_ratio['reorder_ratio'],
    bins=30,
    edgecolor='black',
)

# Etiquetas y título
plt.title('Distribución de Proporciones de Productos Reordenados')
plt.xlabel('Proporción de Reordenes')
plt.ylabel('Cantidad de Productos')
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Mostrar la gráfica
plt.tight_layout()
plt.show()
```

    - Para calcular la proporción de productos creamos primero una lista con tdos los productos que al menos hayan sido reordenados 1 vez y lo dividimos entre el total de ordenes en las que aparezcan, eso genera un valor float entre 0 y 1 donde mientras más cercano sea a 1 es que ese producto es altamante demandado a partir de su primera compra mientras que por el contrario si es 0 es practicamente un hecho que los clientes los compran solo 1 vez. Existen más de 1800 productos con un radio de 1, sin embargo si calulamos la media es de 0.38 debido a que la mayor carga de productos tienden a 0 siendo los dos primeros cuartiles de 0.16 y 0.4 respectivamente y más de 8000 productos que practicamente nunca son reordenados.

### [C4] Para cada cliente, ¿qué proporción de sus productos ya los había pedido?

```python
# Combinar las tablas de pedidos y productos
orders_products = pd.merge(order_products, orders[['order_id', 'user_id']], on='order_id')

# Calcular el total de productos pedidos por cada cliente
user_total_orders = orders_products.groupby('user_id')['product_id'].count().reset_index()
user_total_orders.columns = ['user_id', 'total_products']

# Calcular el total de productos reordenados por cada cliente
user_reordered = orders_products[orders_products['reordered'] >= 1]
user_reordered_counts = user_reordered.groupby('user_id')['product_id'].count().reset_index()
user_reordered_counts.columns = ['user_id', 'reordered_products']

# Combinar los datos
user_reorder_ratio = pd.merge(
    user_total_orders,
    user_reordered_counts,
    on='user_id',
    how='left'
)

# Rellenar valores NaN (clientes que no reordenaron nada)
user_reorder_ratio['reordered_products'] = user_reorder_ratio['reordered_products'].fillna(0)

# Calcular la proporción
user_reorder_ratio['reorder_ratio'] = (
    user_reorder_ratio['reordered_products'] / user_reorder_ratio['total_products']
)

# Mostrar el resultado
#print(user_reorder_ratio)
user_reorder_ratio.describe()
```

```python
plt.hist(user_reorder_ratio['reorder_ratio'], bins=20, edgecolor='black')

# Etiquetas y título
plt.title('Distribución de la Proporción de Productos Reordenados por Cliente')
plt.xlabel('Proporción de Productos Reordenados')
plt.ylabel('Cantidad de Clientes')

# Mostrar la gráfica
plt.tight_layout()
plt.show()
```

    - Ahora nos encontramos con una distribución muy similar a la gráfica anterior sobre la distribución de productos reordenados, sin embargo aquí la comparativa de reorden se genera al dividir la cantidad de productos que un usuario reordena un producto entre el total de usuarios que realizan esta acción. Si observamos la distribución y calculamos la media y los cuartiles aquí notamos que la distribución es casi perfecta teniendo una media de 0.49 y cuartiles de .27, .50, .72 y 1. Por ultimo lo que podemos observar es la alta diferencia entre el valor 0 con más de 16000 clientes y el 1 con cerca de 9000, estos dos valores al igual que en la gráfica anterior se encuentran en alta disparidad.
